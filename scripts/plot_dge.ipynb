```python
import os
os.environ["CUDA_VISIBLE_DEVICES"] = "7"

import torch
from matplotlib import pyplot as plt

from torchtitan.experiments.kernels.blockwise_fp8.differentiable_gradient_estimator import (
    dge_fwd,
    dge_bwd,
    BREAK_POINTS,
    INTERVALS,
)
dtype = torch.float8_e4m3fnuz
# q = torch.randn(64, 64)
# dge_bwd(q, dtype)
```

```python
print(BREAK_POINTS[dtype].shape)
print(INTERVALS[dtype].shape)
```

```python

x = torch.arange(1, 1.25, 1e-3)
#x = torch.arange(-240, 240, 1e-3)
y = dge_fwd(x, dtype)
qx = x.to(dtype).to(torch.float32)

plt.plot(x.numpy(), x.numpy(), label="STE")
plt.plot(x.numpy(), qx.numpy(), label="Hard")
plt.plot(x.numpy(), y.numpy(), label="DGE Forward")
plt.show()
```

```python
dy = dge_bwd(y, dtype)
plt.plot(x.numpy(), dy.numpy(), label="DGE Backward")
plt.plot(x.numpy(), torch.ones_like(x).numpy(), label="STE Backward")
plt.show()
```
